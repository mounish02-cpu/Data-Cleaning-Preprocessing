```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("Libraries loaded successfully.")
```

```python
df = pd.read_csv("/content/raw_dataset.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("========== BEFORE CLEANING ==========")

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Detailed dataset inspection

print("========== DATASET INSPECTION ==========")

# First 5 records
print("\nFirst 5 Records:")
display(df.head())

# Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)

# Duplicate records
duplicate_count = df.duplicated().sum()
print("\nDuplicate Records:", duplicate_count)

# Unique values in categorical columns
categorical_columns = [
    "category",
    "payment_method",
    "region",
    "returned",
    "customer_gender"
]

print("\nUnique Values in Categorical Columns:")
for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
# ==========================================
# OUTLIER AND CONSISTENCY CHECK
# ==========================================

numeric_columns = [
    "price",
    "discount",
    "quantity",
    "delivery_time_days",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "customer_age"
]

print("========== OUTLIER ANALYSIS ==========")

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    print(f"\n{col}")
    print("Lower Bound:", round(lower_bound, 2))
    print("Upper Bound:", round(upper_bound, 2))
    print("Outliers:", outliers)

print("\n========== CONSISTENCY CHECK ==========")

print("Negative prices:", (df["price"] < 0).sum())
print("Invalid discounts:", ((df["discount"] < 0) | (df["discount"] > 1)).sum())
print("Invalid quantities:", (df["quantity"] <= 0).sum())
print("Invalid delivery days:", (df["delivery_time_days"] < 0).sum())
print("Invalid shipping costs:", (df["shipping_cost"] < 0).sum())
print("Invalid customer ages:", ((df["customer_age"] < 18) | (df["customer_age"] > 100)).sum())
```

```python
# ==========================================
# DATA CLEANING AND PREPROCESSING
# ==========================================

clean_df = df.copy()

print("========== STARTING CLEANING ==========")

# 1. Standardize column names
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

# 2. Remove completely duplicate records
before_duplicates = len(clean_df)
clean_df = clean_df.drop_duplicates()
duplicates_removed = before_duplicates - len(clean_df)

print("Duplicate records removed:", duplicates_removed)

# 3. Convert date column to proper datetime
clean_df["order_date"] = pd.to_datetime(
    clean_df["order_date"],
    errors="coerce"
)

# 4. Convert numeric columns to numeric data types
numeric_columns = [
    "price",
    "discount",
    "quantity",
    "delivery_time_days",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "customer_age"
]

for col in numeric_columns:
    clean_df[col] = pd.to_numeric(
        clean_df[col],
        errors="coerce"
    )

# 5. Handle missing values
# Numerical columns -> median
for col in numeric_columns:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].median()
    )

# Categorical columns -> mode
categorical_columns = [
    "category",
    "payment_method",
    "region",
    "returned",
    "customer_gender"
]

for col in categorical_columns:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].mode()[0]
    )

# 6. Apply business-rule validation
clean_df = clean_df[
    (clean_df["price"] >= 0) &
    (clean_df["discount"] >= 0) &
    (clean_df["discount"] <= 1) &
    (clean_df["quantity"] > 0) &
    (clean_df["delivery_time_days"] >= 0) &
    (clean_df["shipping_cost"] >= 0) &
    (clean_df["customer_age"] >= 18) &
    (clean_df["customer_age"] <= 100)
].copy()

# 7. Handle extreme monetary/profit outliers using IQR capping
# We do NOT cap quantity or delivery time because those values
# can be legitimate business observations.

outlier_columns = [
    "price",
    "total_amount",
    "shipping_cost",
    "profit_margin"
]

for col in outlier_columns:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    clean_df[col] = clean_df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Cleaning completed successfully!")
print("Cleaned dataset shape:", clean_df.shape)
```

```python
# ==========================================
# FEATURE ENGINEERING
# ==========================================

print("========== FEATURE ENGINEERING ==========")

# Extract year and month from order date
clean_df["order_year"] = clean_df["order_date"].dt.year
clean_df["order_month"] = clean_df["order_date"].dt.month
clean_df["order_month_name"] = clean_df["order_date"].dt.month_name()

# Calculate profit amount
clean_df["profit_amount"] = (
    clean_df["total_amount"] - clean_df["shipping_cost"]
)

# Calculate profit margin percentage
clean_df["calculated_profit_margin"] = (
    clean_df["profit_amount"] / clean_df["total_amount"] * 100
)

# Handle any division-by-zero results
clean_df["calculated_profit_margin"] = (
    clean_df["calculated_profit_margin"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

print("Feature engineering completed successfully!")

print("\nNew columns added:")
print([
    "order_year",
    "order_month",
    "order_month_name",
    "profit_amount",
    "calculated_profit_margin"
])

print("\nSample of engineered features:")
display(
    clean_df[
        [
            "order_date",
            "order_year",
            "order_month",
            "order_month_name",
            "total_amount",
            "shipping_cost",
            "profit_amount",
            "calculated_profit_margin"
        ]
    ].head(10)
)
```

```python
# ==========================================
# BEFORE vs AFTER CLEANING COMPARISON
# ==========================================

print("========== BEFORE vs AFTER CLEANING ==========")

comparison = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Missing Values",
        "Duplicate Records",
        "Invalid Prices",
        "Invalid Discounts",
        "Invalid Quantities",
        "Invalid Delivery Days",
        "Invalid Shipping Costs",
        "Invalid Customer Ages"
    ],

    "Before Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        (df["price"] < 0).sum(),
        ((df["discount"] < 0) | (df["discount"] > 1)).sum(),
        (df["quantity"] <= 0).sum(),
        (df["delivery_time_days"] < 0).sum(),
        (df["shipping_cost"] < 0).sum(),
        ((df["customer_age"] < 18) | (df["customer_age"] > 100)).sum()
    ],

    "After Cleaning": [
        clean_df.shape[0],
        clean_df.shape[1],
        clean_df.isnull().sum().sum(),
        clean_df.duplicated().sum(),
        (clean_df["price"] < 0).sum(),
        ((clean_df["discount"] < 0) | (clean_df["discount"] > 1)).sum(),
        (clean_df["quantity"] <= 0).sum(),
        (clean_df["delivery_time_days"] < 0).sum(),
        (clean_df["shipping_cost"] < 0).sum(),
        ((clean_df["customer_age"] < 18) | (clean_df["customer_age"] > 100)).sum()
    ]
})

display(comparison)
```

```python
# ==========================================
# FINAL DATA QUALITY VERIFICATION
# ==========================================

print("========== FINAL DATA QUALITY CHECK ==========")

print("Final Shape:", clean_df.shape)
print("Total Missing Values:", clean_df.isnull().sum().sum())
print("Total Duplicate Rows:", clean_df.duplicated().sum())

print("\nFinal Data Types:")
print(clean_df.dtypes)

print("\nFinal Dataset Preview:")
display(clean_df.head(10))
```

```python
# ==========================================
# EXPORT CLEAN DATASET
# ==========================================

# Save the cleaned dataset as CSV
clean_df.to_csv(
    "/content/clean_dataset.csv",
    index=False
)

print("Clean dataset exported successfully!")
print("File name: clean_dataset.csv")
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
```
